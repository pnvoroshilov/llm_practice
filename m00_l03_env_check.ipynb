```python
import platform
import sys
import torch

print("Python:", sys.version.split()[0], "| ОС:", platform.system())
print("PyTorch:", torch.__version__)

if torch.cuda.is_available():            # видеокарта NVIDIA
    device = "cuda"
    p = torch.cuda.get_device_properties(0)
    print("GPU:", p.name, "| память:", round(p.total_memory / 1024**3, 1), "ГиБ")
elif torch.backends.mps.is_available():  # Mac на Apple Silicon
    device = "mps"
    print("Apple GPU (MPS), память общая с системой")
else:
    device = "cpu"
    print("GPU не найден, работаем на CPU")

x = torch.randn(1000, 1000, device=device)
y = x @ x                                # матричное умножение на выбранном устройстве
print("Матрица на:", y.device, "| форма:", tuple(y.shape))
```
